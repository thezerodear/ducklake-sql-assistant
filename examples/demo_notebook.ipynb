# 🦆 DuckLake SQL Assistant Demo Notebook

This notebook connects to `mock_ducklake.duckdb` and demonstrates SQL syntax highlighting and auto-completion inside triple quotes.

```python
import duckdb

# Connect to local mock DuckDB database
con = duckdb.connect('mock_ducklake.duckdb')
print("Connected to mock DuckDB!")
con.sql("SHOW TABLES;").show()
```

```python
# Test 1: Autocomplete after alias 'u.' or 'o.'
# Try typing 'u.' or 'o.' in the SELECT clause
df = con.sql("""
    SELECT 
        u.
        u.username,
        u.email,
        o.amount,
        o.order_status
    FROM lake_users u
    JOIN lake_orders o ON u.user_id = o.user_id
    WHERE u.status = 'ACTIVE'
""").df()
print(df)
```

```python
# Test 2: Autocomplete table names after 'FROM '
# Try deleting 'lake_events' and typing 'FROM ' to see table suggestions
con.sql("""
    SELECT 
        event_id,
        event_type,
        created_at
    FROM lake_events
""")
```

```python
import duckdb

# 1. สร้าง session ของ DuckDB
con = duckdb.connect()

# 2. ติดตั้งและโหลด extension ducklake กับ postgres
con.execute("INSTALL ducklake; INSTALL postgres;")
con.execute("LOAD ducklake; LOAD postgres;")

# 3. กำหนดค่า Connection ไปยัง PostgreSQL Catalog (port 5439) และ Data Path
pg_conn_str = "dbname=ducklake_catalog host=localhost port=5439 user=postgres password=ducklake123"
data_path = "C:/Users/theze/.gemini/antigravity/scratch/ducklake-mockup/data/"

# 4. ทำการ ATTACH DuckLake เข้ากับ DuckDB
con.execute(f"""
    ATTACH 'ducklake:postgres:{pg_conn_str}' 
    AS lake (DATA_PATH '{data_path}');
""")

# 5. สลับไปใช้งาน database 'lake'
con.execute("USE lake;")

# 6. ตรวจสอบรายชื่อตาราง (จะขึ้น customers, orders, products)
con.sql("SHOW TABLES;").show()
```

```python
import duckdb

# 1. Connect DuckDB
con = duckdb.connect()

# 2. Install & load ducklake and postgres extensions
con.execute("INSTALL ducklake; INSTALL postgres;")
con.execute("LOAD ducklake; LOAD postgres;")

# 3. Attach DuckLake Catalog
con.execute("""
    ATTACH 'ducklake:postgres:dbname=ducklake_catalog host=localhost port=5439 user=postgres password=ducklake123' 
    AS lake;
""")

# 4. Use database
con.execute("USE lake;")
con.sql("SHOW TABLES;").show()
```
