# 🦆 DuckLake SQL Assistant Demo Notebook

Connected to DuckLake over PostgreSQL Catalog (localhost:5439) with local Parquet storage.

```python
import duckdb

# 1. Initialize DuckDB
con = duckdb.connect()

# 2. Load DuckLake and PostgreSQL extensions
con.execute("INSTALL ducklake; INSTALL postgres;")
con.execute("LOAD ducklake; LOAD postgres;")

# 3. Attach DuckLake over PostgreSQL Catalog
con.execute("""
    ATTACH 'ducklake:postgres:dbname=ducklake_catalog host=localhost port=5439 user=postgres password=ducklake123'
    AS lake (DATA_PATH 'C:/Users/theze/.gemini/antigravity/scratch/ducklake-mockup/data/');
""")

con.execute("USE lake;")
print("Connected to DuckLake!")
con.sql("SHOW TABLES;").show()
```

```python
# Test 1: Autocomplete after alias 'c.' or 'o.'
# Try typing 'c.' or 'o.' in the SELECT clause
con.sql("""
    SELECT 
        c.customer_id,
        c.name,
        c.email,
        o.order_id,
        o.quantity,
        o.status
    FROM customers c
    JOIN orders o ON c.customer_id = o.customer_id
    WHERE o.status = 'COMPLETED'
""").show()
```

```python
# Test 2: DuckDB COLUMNS() dynamic column selection
# Notice syntax highlighting for COLUMNS, EXCLUDE, and lambda '->'
con.sql("""
    SELECT 
        COLUMNS('^customer_.*'),
        COLUMNS(* EXCLUDE (signup_date))
    FROM customers
""").show()
```
